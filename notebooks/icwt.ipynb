```python
import numpy as np

from piwavelet.transforms.cwt import cwt
from piwavelet.transforms.icwt import icwt
from piwavelet.wavelets.morlet import Morlet
```

```python
# ---------------------------------------------------------
# Synthetic signal
# ---------------------------------------------------------

dt = 0.01

t = np.arange(0, 10, dt)

signal = (
    np.sin(2 * np.pi * 5 * t)
    + 0.5 * np.sin(2 * np.pi * 15 * t)
)
```

```python
# ---------------------------------------------------------
# Wavelet transform
# ---------------------------------------------------------

wavelet = Morlet()

cwt_result = cwt(
    signal=signal,
    dt=dt,
    wavelet=wavelet,
)
```

```python
# ---------------------------------------------------------
# Inverse transform
# ---------------------------------------------------------

reconstructed = icwt(
    coefficients=cwt_result.coefficients,
    scales=cwt_result.scales,
    dt=dt,
    dj=1 / 12,
    wavelet=wavelet,
)
```

```python
# ---------------------------------------------------------
# Reconstruction error
# ---------------------------------------------------------

rms_error = np.sqrt(
    np.mean((signal - reconstructed) ** 2)
)

correlation = np.corrcoef(
    signal,
    reconstructed,
)[0, 1]

print(f"RMS error: {rms_error:.6e}")
print(f"Correlation: {correlation:.12f}")
```
